```python
import pandas as pd
import numpy as np

columns = [
    'age', 'sex', 'cp', 'trestbps', 'chol', 
    'fbs', 'restecg', 'thalach', 'exang', 
    'oldpeak', 'slope', 'ca', 'thal', 'target']
```

```python
df = pd.read_csv('heart+disease/processed.cleveland.data', names=columns, na_values='?')

print("--- First 5 Rows ---")
display(df.head())

print(f"\nDataset Shape: {df.shape}")

print("\n--- Data Types ---")
print(df.dtypes)
```

```python
missing_values = df.isnull().sum()
print("--- Missing Values per Column ---")
print(missing_values[missing_values > 0])
```

```python
ca_median = df['ca'].median()
df['ca'] = df['ca'].fillna(ca_median)

thal_mode = df['thal'].mode()[0]
df['thal'] = df['thal'].fillna(thal_mode)

print("Missing values after imputation:")
print(df[['ca', 'thal']].isnull().sum())
```

```python
df['sex'] = df['sex'].astype('category')
df['cp'] = df['cp'].astype('category')

print("--- Updated Data Types ---")
print(df[['sex', 'cp']].dtypes)
```

```python
print("--- Cholesterol Descriptive Stats ---")
print(df['chol'].describe())
```

```python
chol_cap = df['chol'].quantile(0.99)

df.loc[df['chol'] > chol_cap, 'chol'] = chol_cap

print(f"Max cholesterol after capping: {df['chol'].max()}")
```

```python
duplicates_count = df.duplicated().sum()
print(f"Total duplicates found: {duplicates_count}")

df = df.drop_duplicates()

print(f"Dataset Shape after duplicate removal: {df.shape}")
```

```python
print("--- FINAL VALIDATION CHECK ---")
print(f"1. Any Missing Values left? {'Yes' if df.isnull().values.any() else 'No'}")
print(f"2. Current Shape (No duplicates): {df.shape}")
print("\n3. Cleaned Data Types:")
print(df.dtypes)
```
