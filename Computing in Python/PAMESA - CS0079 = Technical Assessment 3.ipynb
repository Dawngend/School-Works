## Part A

```python
class Pokemon:
    """
    Base class representing a generic Pokémon.
    
    Attributes:
        category (str): A class-level attribute shared across all instances.
        name (str): The name of the Pokémon.
        level (int): The current level of the Pokémon (must be a positive integer).
    """
    
    # Class Attribute
    category = "Pocket Monster"
    
    def __init__(self, name: str, level: int):
        self.name = name
        # We initialize the private variable first, then use the property setter 
        # to validate the incoming level parameter safely.
        self._level = 1 
        self.level = level

    @property
    def level(self) -> int:
        """Getter for the level attribute."""
        return self._level

    @level.setter
    def level(self, value):
        """
        Setter for the level attribute with validation.
        Ensures the input is a positive whole number. Handles invalid inputs gracefully.
        """
        try:
            val = int(value)
            if val > 0:
                self._level = val
            else:
                print(f"[!] Invalid level '{value}' for {self.name}. Must be positive. Defaulting to 1.")
                self._level = 1
        except (ValueError, TypeError):
            print(f"[!] Invalid input type '{value}' for {self.name}'s level. Defaulting to 1.")
            self._level = 1

    def describe(self):
        """Prints a descriptive string of the Pokémon's current state."""
        print(f"This Pokémon is named {self.name} and is at level {self.level}.")
```

## Part B


```python
class FirePokemon(Pokemon):
    """
    Child class representing a Fire-type Pokémon. Inherits from Pokemon.
    
    Additional Attributes:
        fire_power (int): The elemental strength of the Pokémon (must be positive).
    """
    
    def __init__(self, name: str, level: int, fire_power: int):
        # Call the parent class constructor to handle name and level
        super().__init__(name, level)
        
        # Initialize and validate fire_power
        self._fire_power = 1
        self.fire_power = fire_power

    @property
    def fire_power(self) -> int:
        """Getter for the fire_power attribute."""
        return self._fire_power

    @fire_power.setter
    def fire_power(self, value):
        """
        Setter for the fire_power attribute with validation.
        Ensures the input is a positive whole number.
        """
        try:
            val = int(value)
            if val > 0:
                self._fire_power = val
            else:
                print(f"[!] Invalid fire power '{value}' for {self.name}. Must be positive. Defaulting to 1.")
                self._fire_power = 1
        except (ValueError, TypeError):
            print(f"[!] Invalid input type '{value}' for {self.name}'s fire power. Defaulting to 1.")
            self._fire_power = 1

    def attack_power(self) -> int:
        """Calculates and returns the total attack power."""
        return self.level * self.fire_power
```

## Part C

```python
class FirePokemon(Pokemon):
    """
    Child class representing a Fire-type Pokémon. Inherits from Pokemon.
    This version includes the overridden describe() method for Part C.
    """
    
    def __init__(self, name: str, level: int, fire_power: int):
        super().__init__(name, level)
        
        self._fire_power = 1
        self.fire_power = fire_power

    @property
    def fire_power(self) -> int:
        return self._fire_power

    @fire_power.setter
    def fire_power(self, value):
        try:
            val = int(value)
            if val > 0:
                self._fire_power = val
            else:
                print(f"[!] Invalid fire power '{value}' for {self.name}. Must be positive. Defaulting to 1.")
                self._fire_power = 1
        except (ValueError, TypeError):
            print(f"[!] Invalid input type '{value}' for {self.name}'s fire power. Defaulting to 1.")
            self._fire_power = 1

    def attack_power(self) -> int:
        return self.level * self.fire_power

    # === THE OVERRIDDEN METHOD ===
    def describe(self):
        """Overrides the base describe method to include Fire-type specifics."""
        print(f"Fire-type Pokémon: {self.name}, Level {self.level}, Fire Power {self.fire_power}.")


# --- Part C Test Case ---
if __name__ == "__main__":
    print("--- PART C: Overridden Method Test ---")
    charizard = FirePokemon("Charizard", 30, 10)
    charizard.describe()
```

```python
class WaterPokemon(Pokemon):
    """
    Child class representing a Water-type Pokémon. Inherits from Pokemon.
    
    Additional Attributes:
        water_power (int): The elemental strength of the Pokémon (max 100).
    """
    
    def __init__(self, name: str, level: int, water_power: int):
        # Call the parent class constructor to handle name and level
        super().__init__(name, level)
        
        # Initialize and validate water_power
        self._water_power = 1
        self.water_power = water_power

    @property
    def water_power(self) -> int:
        """Getter for the water_power attribute."""
        return self._water_power

    @water_power.setter
    def water_power(self, value):
        """
        Setter for water_power with validation and a strict cap.
        Ensures input is positive, and caps any value over 100 down to 100.
        """
        try:
            val = int(value)
            if val > 100:
                print(f"[*] Capping {self.name}'s water power from {val} to the maximum of 100.")
                self._water_power = 100
            elif val > 0:
                self._water_power = val
            else:
                print(f"[!] Invalid water power '{value}' for {self.name}. Must be positive. Defaulting to 1.")
                self._water_power = 1
        except (ValueError, TypeError):
            print(f"[!] Invalid input type '{value}' for {self.name}'s water power. Defaulting to 1.")
            self._water_power = 1

    def attack_power(self) -> int:
        """Calculates and returns the total attack power."""
        return self.level * self.water_power
```

```python
def show_pokemon_info(pokemon):
    """
    Accepts any object derived from the Pokemon base class.
    Demonstrates polymorphism by calling shared methods that may 
    behave differently depending on the specific child class passed in.
    """
    pokemon.describe()
    
    # We use a try-except block here just in case the base Pokemon class 
    # (which doesn't have an attack_power method in Part A) is passed by mistake.
    try:
        print(f"Attack Power: {pokemon.attack_power()}\n")
    except AttributeError:
        print("[!] This base Pokémon does not have an attack power calculation.\n")
```
