# Laboratory Activity

```python
import numpy as np

if 'movie_database' not in globals():
    movie_database = {
        "1": {"title": "Movie A: The Matrix", "seats": np.arange(1, 101).reshape(10, 10)},
        "2": {"title": "Movie B: Inception", "seats": np.arange(1, 101).reshape(10, 10)},
        "3": {"title": "Movie C: Interstellar", "seats": np.arange(1, 101).reshape(10, 10)}
    }

def displaySeats(movie):
    print(f"\n--- {movie['title']} Seating ---")
    print("=" * 55)
    print("                       SCREEN                      ")
    print("=" * 55)
    
    seats = movie["seats"]
    for row in seats:
        for seat in row:
            if seat == 0:
                # Perfectly aligns the [X] with the numbers
                print(f"{'[X]':>4} ", end=" ") 
            else:
                print(f"{seat:4d} ", end=" ")
        print() 
    print("=" * 55)
    print("Legend: [X] = Reserved\n")

def ticketReservation(movies):
    print("\n--- Available Movies ---")
    for key, val in movies.items():
        print(f"[{key}] {val['title']}")
        
    m_choice = input("Select a movie (1-3): ")
    if m_choice in movies:
        displaySeats(movies[m_choice])
        try:
            seat_num = int(input("Enter seat number to reserve (1-100): "))
            if 1 <= seat_num <= 100:
                row = (seat_num - 1) // 10
                col = (seat_num - 1) % 10
                
                if movies[m_choice]["seats"][row][col] == 0:
                    print("\n[!] Sorry, that seat is already reserved.")
                else:
                    movies[m_choice]["seats"][row][col] = 0
                    print(f"\n[*] Success! Seat {seat_num} reserved for {movies[m_choice]['title']}.")
                    
                    # NEW: Instantly display the seats again so you can see your [X]!
                    displaySeats(movies[m_choice])
            else:
                print("\n[!] Invalid seat number.")
        except ValueError:
            print("\n[!] Please enter a valid number.")
    else:
        print("\n[!] Invalid movie selection.")

def cancelReservation(movies):
    print("\n--- Available Movies ---")
    for key, val in movies.items():
        print(f"[{key}] {val['title']}")
        
    m_choice = input("Select the movie for your reservation (1-3): ")
    if m_choice in movies:
        try:
            seat_num = int(input("Enter seat number to cancel (1-100): "))
            if 1 <= seat_num <= 100:
                row = (seat_num - 1) // 10
                col = (seat_num - 1) % 10
                
                if movies[m_choice]["seats"][row][col] == 0:
                    movies[m_choice]["seats"][row][col] = seat_num
                    print(f"\n[*] Success! Reservation for seat {seat_num} cancelled.")
                    
                    # NEW: Instantly display the seats to show the number is back!
                    displaySeats(movies[m_choice])
                else:
                    print("\n[!] That seat is not currently reserved.")
            else:
                print("\n[!] Invalid seat number.")
        except ValueError:
            print("\n[!] Please enter a valid number.")
    else:
        print("\n[!] Invalid movie selection.")

def main():
    while True:
        print("\n" + "*"*30)
        print("   MOVIE TICKET SYSTEM")
        print("*"*30)
        print("1. Display Seats")
        print("2. Reserve a Ticket")
        print("3. Cancel a Reservation")
        print("4. Exit")
        
        choice = input("Enter your choice (1-4): ")
        
        if choice == '1':
            print("\n--- Available Movies ---")
            for key, val in movie_database.items():
                print(f"[{key}] {val['title']}")
            m_choice = input("Select a movie to view (1-3): ")
            if m_choice in movie_database:
                displaySeats(movie_database[m_choice])
            else:
                print("\n[!] Invalid movie selection.")
                
        elif choice == '2':
            ticketReservation(movie_database)
            
        elif choice == '3':
            cancelReservation(movie_database)
            
        elif choice == '4':
            print("\nExiting system. Goodbye!")
            break
            
        else:
            print("\n[!] Invalid choice. Please select 1-4.")

if __name__ == "__main__":
    main()
```
