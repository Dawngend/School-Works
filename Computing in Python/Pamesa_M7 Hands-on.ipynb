```python
x = np.linspace(-10,10,20)
y = x**2
z = x*10
plt.plot(x,y,"-.", label ="x^2")
plt.plot(x,z,":", label="x*10")
plt.title("Plot of Some Functions from -10 to 10")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(loc=1)
plt.show()
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt
random.seed(20)
x = random.choices(range(100), k=20)
y = np.linspace(0,100,20)
plt.scatter(x,y, color="red")
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
x = np.linspace(-10,10,20)
y = x**2
z = x*10
plt.plot(x,y,"-.", label="x^2")
plt.plot(x,z, ":", label="x*10")
plt.title("Plot of Some Functions from -10 to 10")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.ylim(-120, 120)
plt.xlim(-15, 15)
plt.locator_params(axis='x', nbins=20)
plt.locator_params(axis='y', nbins=20)
plt.grid()
plt.legend(loc=0)
plt.show()
```

```python
%matplotlib inline

x = np.arange(20)
y = np.random.randint(1,100,20).cumsum()

plt.figure(figsize=(14,8))

plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
```

```python
plt.subplot(2,2,2)
plt.scatter(x,y, color="red")
plt.title("Scatter")
plt.xlabel("X")
plt.ylabel("Y")

plt.subplot(2, 2, 3)
plt.bar(x,y, color ="green")
plt.title("Bar")
plt.xlabel("X")
plt.ylabel("Y")

plt.subplot(2,2,4)
plt.hist(y, bins=20, color="cyan" , edgecolor='black')
plt.title("Histogram (number of bins = 20)")
plt.ylabel("Frequency")
plt.tight_layout()
```

```python
%matplotlib inline

x = np.arange(20)
y = np.random.randint(1,100,20).cumsum()

plt.figure(figsize=(14,8))

plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()

plt.subplot(2,2,4)
plt.hist(y, bins=20, color="cyan" , edgecolor='black')
plt.title("Histogram (nbins = 20)")
plt.ylabel("Frequency")
plt.tight_layout()
```
